```python
# leer de forma estructurada el fichero alumnos.txt y sacar la informacíón que hay en este
with open("alumnos.txt", "rt") as fpAlumnos:
    for alumno in fpAlumnos:
        # para eliminar espacios (espacios en blanco, tabulador, carros de retorno) usamos strip()
        print(alumno.strip())
```

```python
# expresión regular: gramática que deescribe cadenas texto
# expresíon regular más sencilla: literal (algo que no tiene ningun carácter especial)
# biblioteca re para usar las expresiones regulares
# re.search() y re.match() hacen cosas semejantes
# re.search() localiza una expresión regular y la devuelve si la encuentra

import re
re.search("expresión regular", "El literal es la expresión regular más sencilla") 
```

```python
re.search(".xpresión regular", "El literal es la expresión regular más sencilla") # el punto se puede substituir por cualquier caracter
```

```python
re.search("^.xpresión regular", "El literal es la expresión regular más sencilla") # no coincide por que no encuentra 
```

```python
re.search("^El literal", "El literal es la expresión regular más sencilla") 
```

```python
re.match("El literal", "El literal es la expresión regular más sencilla") 
```

```python
re.search("sencilla$", "antes muerta que sencilla.") # $ obliga a que sea el último carácter de la cadena
```

```python
re.search("sencilla*", "antes muerta que sencill") # * permite repetir carácter 0 o más
```

```python
re.search("sencilla+", "antes muerta que sencillaaa") # + permite repetir carácter 1 o más
```

```python
re.search("sencilla?$", "antes muerta que sencill")  # ? uno o 0
```

```python
re.search("sencilla{3,5}$", "antes muerta que sencillaaaaaa") # solo coincide {m, n}, si no se pone m es 0, si no se pone n es infinito
```

```python
# * = { , }, + = {1, }, ? = {0,1}, {n} = {n, n}
# [] solo permite los caracteres que contenga
re.search("sencill[aoi]", "antes muerta que sencille") 
```

```python
# [^ ] el ^ sirve para excluir 
re.search("sencill[^eu]", "antes muerta que sencill") 
```

```python
re.search("sencill(e)", "antes muerta que sencill(e)") 
```

```python
re.search("sencill\\(e\\)", "antes muerta que sencill(e)") 
```

```python
re.search("https:\\\\\\\\upc.edu", "https:\\\\upc.edu\\eseiaat")
```

```python
re.search(r"https:\\\\upc.edu", r"https:\\upc.edu\eseiaat")
```
