## Cadenas con formato

```python
"Una cadena de texto con {1} y {0}".format("algo", 3) # argumentos posicionales
```

```python
"Una cadena de texto con {algo} y {tres}".format(algo = "algo", tres = 3) # argumentos por clave
```

```python
"Una cadena de texto con {algo} y {tres}".format(algo = "algo", tres = 793090435.5498354)
```

```python
"Una cadena de texto con {algo} y {tres:.2f}".format(algo = "algo", tres = 793090435.5498354) # precisión
```

```python
"Una cadena de texto con {algo} y {tres:32.2f}".format(algo = "algo", tres = 793090435.5498354) # anchura (espacio que ocupa el objeto)
```

```python
"Una cadena de texto con {algo} y {tres:32_.2f}".format(algo = "algo", tres = 793090435.5498354) # agrupa para facilitar la lectura
```

```python
"Una cadena de texto con {algo} y {tres:<32_.2f}".format(algo = "algo", tres = 793090435.5498354) # alineado con los caracteres '<', '>', '^'
```

```python
"Una cadena de texto con {algo} y {tres:-^32_.2f}".format(algo = "algo", tres = 793090435.5498354) # Si delante de los caracteres de alineado añadimos un caracter, el espacio vacío se rellenará con dicho caracter
```

```python
"{numero}".format(numero = 5156541)
```

```python
"{numero:o}".format(numero = 5156541) # número en ocotal
```

```python
"{numero:#o}".format(numero = 5156541) # número en ocotal con la información indicando que el número está en ocotal
```

```python
"{numero:b}".format(numero = 5156541) # número en binaro
```

```python
"{numero:#_b}".format(numero = 5156541) # número en binario con la etiqueta del binario y serparación típica de números binarios
```

```python
numero = 6564
```

```python
"{numero:{alternativo}{agrupa}{tipo}}".format(numero=5156541, alternativo="", agrupa="" if numero < 100 else "_" , tipo="b" ) # usamos condicional ternario
```

```python
from datetime import datetime as dt 
```

```python
dt.now()
```

```python
"Ahora son las {ahora}".format(ahora=dt.now())
```

```python
"Ahora son las {ahora:%I %p y %M minutos del %A, %d de %B de %Y}".format(ahora=dt.now())
```

```python
import locale
locale.setlocale(locale.LC_ALL, "es_ES.UTF8")
```

```python
"Ahora son las {ahora:%H y %M minutos del %A, %d de %B de %Y}".format(ahora=dt.now())
```

```python
locale.setlocale(locale.LC_ALL, "ca_ES.UTF8")
```

```python
"Ara son les {ahora:%H y %M minuts del %A, %d d'%B de %Y}".format(ahora=dt.now())
```

```python
"numero = {numero}".format(numero = numero)
```

```python
# anteriormente hemos repetido 4 veces la palabra número, nos interesa reducir este número
f"numero={numero}"
```

```python
f"numero={numero:·^40}"
```

```python
f"numero={numero:·^40.1f}" # lo convierte a real y añade un decimal
```

```python
import numpy as np
numero = numero ** 2
f"numero = {numero:·^{int(np.floor(np.log10(numero)+1))}}"
```

```python
f"numero = {numero:·^{int(np.floor(np.log10(numero)+7))}}"
```

```python
f"numero = {numero}"
```

```python
f"np.cos(numero**2) = {np.cos(numero**2)}"
```

```python
f"{np.cos(numero**2) = :.4f}" # con el = nos aseguramos que la expresion coincida con el resultado y nos ahorramos escribir dos veces la expresión, corriendo el peligro de que nos podamos equivocar
```
