```python
st.pack("@hf", entero, real) # Al generar, si usamos la misma maquia usamos el @ que indica que se usa el formato nativo de la maquina, no hay conversiones, es más eficiente
```

```python
# Escribimos estructura de bytes dentro de un fichero:
# gestor de contexto (ENTRA EXAMEN)

with open("fichero.bin", "wb") as fpbin: # wb para escritura binaria, para leer rb (lectura binaria) o rt (lectura de texto)
    formato = "@hf"
    datos = st.pack(formato, entero, real)
    fpbin.write(datos)
```

```python
with open("fichero.bin","rb") as fpbin:
    formato = "@hf"
    datos = fpbin.read()
    entero, real = st.unpack(formato, datos)

print(f"{entero = } \t {real = }")
```

En el char tipo pascal el primer byte indica la longitud de la cadena. No hay basura, ecofriendly
Vamos a tener que esatar obligados a usar el tipo s pues en algunos casos no disponemos de ese byte extra del tipo pascal

```python
st.pack("@iiii", 555, 222, 666, 888)
```

```python
cadena = st.pack("@6s9s12s", b"muy larga", b"muy justa", b"muy corta")
larga, justa, corta = st.unpack("@6s9s12s", cadena)
print(larga, justa, corta, sep =  "\n") 
```

```python
cadena = st.pack("@6p9p12p", b"muy larga", b"muy justa", b"muy corta")
larga, justa, corta = st.unpack("@6p9p12p", cadena)
print(larga, justa, corta, sep =  "\n") # justa se ve truncada debido al byte extra que usa el formato pascal
```

```python
alumnos = [[1, "Monsterrat Cuevas López".encode(), 9],
           [2, "Imelda Run Montes Martín".encode(), 6.5],
           [3, "Oriol Jiménez Garrich".encode(), 4.01],
          ] # la última coma es muy útil para añadir o quitar el útlimo alumno
```

```python
formato = "<I120sf" # portable little endian
```

```python
with open ("alumnes.bin", "wb") as fpAlumnos:
 for alumno in alumnos:
     datos = st.pack(formato, *alumno) # *alumno es lo mismo que hacer alumno[0], alumno[1], alumno[2]
     fpAlumnos.write(datos)
```

```python
with open("alumnes.bin", "rb") as fpAlumnos:
    alumnos = []
    while (datos := fpAlumnos.read(st.calcsize(formato))):
        alumnos.append(st.unpack(formato, datos))
```
