```python
def function(**cosas):
    print(cosas)
    for cosa in cosas:
        print(f"{cosa} -> {cosas[cosa]}") #escribimos clave y contenido
```

```python
function(seat = "Blanco", fiat = "Azul", ford = "Negro")
```

```python
def function(sep = "->", **cosas):
    print(cosas)
    for cosa in cosas:
        print(f"{cosa}{sep}{cosas[cosa]}") #escribimos clave y contenido
```

```python
function(seat = "Blanco", fiat = "Azul", ford = "Negro")
```

```python
function(sep = ":", seat = "Blanco", fiat = "Azul", ford = "Negro")
```
